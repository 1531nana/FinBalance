```python
import pandas as pd

# Selección de las columnas necesarias para el análisis
cols_needed = [
    'P407',
    'T401_3','T401_4','T401_8','T405_2','T405_3','T405_8',
    'T402_3','T406_2','P406A','T405_4',
    'EDAD','GENERO','P704','P705','P708','ESTRATO','REGIÓN','RURALIDAD',
    'T208_1','T208_2','T208_3','T208_5','T208_6','T208_8','T208_9',
    'T208_10','P714','P716',
    'P715','T211_1','T211_2','T211_5',
    'T101_1','T101_2','T101_5','T101_6','T101_7','T101_8','P103','P104',
    'P301',
    'T406_1','T406_3','T406_4',
]

# Cargar el dataset con solo las columnas necesarias
df = pd.read_excel('../data/raw/Encuesta_demanda_2022_microdatos.xlsx',
                   usecols=cols_needed)
df = df[cols_needed]
```

```python
variables_clave = {
    'edad'                      : 'Edad',
    'genero'                    : 'Género',
    'nivel_educativo'           : 'Nivel educativo',
    'ruralidad'                 : 'Ruralidad',
    'rango_ingresos_hogar'      : 'Rango de ingresos',
    'rango_gastos_hogar'        : 'Rango de gastos',
    'prod_cuenta_ahorro'        : 'Tiene cuenta de ahorro',
    'prod_monedero_digital'     : 'Tiene monedero digital',
    'solicito_credito'          : 'Solicitó crédito',
    'comp_tiene_plan_financiero': 'Tiene plan financiero',
    'comp_atraso_pagos'         : 'Atraso en pagos',
    'comp_frecuencia_cumple_plan': 'Frecuencia cumple plan',
}

for col, nombre in variables_clave.items():
    print(f"\n--- {nombre} ---")
    print(df[col].value_counts())
```

```python
n = len(df)

def pct(col, valores):
    count = df[col].isin(valores).sum()
    return count, round(count / n * 100, 1)

print("=== VARIABLE OBJETIVO ===")
c, p = pct('fragilidad_financiera', ['Una semana','Dos semanas'])
print(f"Fragilidad ALTA  (≤2 semanas): {c} ({p}%)")

c, p = pct('fragilidad_financiera', ['Un mes','Tres meses'])
print(f"Fragilidad MEDIA (1-3 meses):  {c} ({p}%)")

c, p = pct('fragilidad_financiera', ['Seis meses','Un año','Más de un año'])
print(f"Fragilidad BAJA  (≥6 meses):   {c} ({p}%)")

ns = df['fragilidad_financiera'].str.contains('NS/NR', na=False).sum()
print(f"NS/NR (excluir antes de entrenar): {ns} ({round(ns/n*100,1)}%)")

print("\n=== PERFIL SOCIOECONÓMICO ===")
print(f"Edad promedio: {df['edad'].mean():.1f} años | Min: {df['edad'].min()} | Max: {df['edad'].max()}")

c, p = pct('genero', ['Femenino'])
print(f"Mujeres: {c} ({p}%)")

c, p = pct('nivel_educativo', ['Ninguno','Primaria','Secundaria'])
print(f"Educación ≤ secundaria: {c} ({p}%)")

c, p = pct('ruralidad', ['Ciudades y aglomerac'])
print(f"Zona urbana: {c} ({p}%)")

c, p = pct('rango_ingresos_hogar', ['Menos de 250.000','Entre 250.001 y 500.000',
                                     'Entre 500.001 y 750.000','Entre 750.001 y 1.000.000'])
print(f"Ingresos < $1.000.000: {c} ({p}%)")

print("\n=== PRODUCTOS FINANCIEROS ===")
c, p = pct('prod_cuenta_ahorro', ['SÍ'])
print(f"Tiene cuenta de ahorro: {c} ({p}%)")

c, p = pct('prod_monedero_digital', ['SÍ'])
print(f"Tiene monedero digital: {c} ({p}%)")

c, p = pct('prod_cdt', ['SÍ'])
print(f"Tiene CDT: {c} ({p}%)")

print("\n=== HÁBITOS FINANCIEROS ===")
c, p = pct('comp_tiene_plan_financiero', ['Sí'])
print(f"Tiene plan financiero: {c} ({p}%)")

c, p = pct('comp_atraso_pagos', ['Siempre','Casi siempre'])
print(f"Atraso frecuente en pagos: {c} ({p}%)")

c, p = pct('comp_atraso_pagos', ['Nunca','Casi nunca'])
print(f"Nunca o casi nunca se atrasa: {c} ({p}%)")

c, p = pct('solicito_credito', ['Sí y fue la primera vez que solicité uno',
                                 'Sí y además tengo otros créditos'])
print(f"Solicitó crédito en el último año: {c} ({p}%)")
```

```python
# Imputar nulos
df['tipo_vinculacion_laboral']       = df['tipo_vinculacion_laboral'].fillna('No aplica')
df['beneficiario_subsidio']          = df['beneficiario_subsidio'].fillna('No informa')
df['producto_financiero_mas_usado']  = df['producto_financiero_mas_usado'].fillna('No informa')
df['comp_frecuencia_cumple_plan']    = df['comp_frecuencia_cumple_plan'].fillna('No tiene plan')

print(df[['tipo_vinculacion_laboral','beneficiario_subsidio',
          'producto_financiero_mas_usado','comp_frecuencia_cumple_plan']].isnull().sum())
```
